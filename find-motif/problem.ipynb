### Opens the file and adds the sequences to a list

```
# Function to open and convert file to list of sequences
def get_seq(filename):
    # String to add the sequence lines onto the string which will be stored in a list
    temp_sequence = ''
    seq_list = [] # List of sequences from the file
    temp_list = []
    # Opens the sequence file
    sequence = open(filename, 'r')

    # Remove the \n's and add to list
    for line in sequence.readlines():
        line = line.strip()
        temp_list.append(line)

    # Combine each sequence into a single string then add to list
    # Ignore the > markers
    for seq in temp_list:
        if seq[0] != '>':
            temp_sequence += seq
        elif seq[0] == '>' and len(temp_sequence) > 0:
            seq_list.append(temp_sequence)
            temp_sequence = ''

    return seq_list
```

```
sequences = get_seq('rosalind_lcsm-6.txt')
longest = ''

for i in range(len(sequences[0])):
    for j in range(i +1, len(sequences[0]) +1):
        substring = sequences[0][i:j]

        if all(substring in sequence for sequence in sequences):
            if len(substring) > len(longest):
                longest = substring

```
