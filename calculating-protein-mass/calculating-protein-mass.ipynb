```python
# Gets dictionary of amino acid: monoisotopic mass from mass table file
mass_table = {}
numstring = ''
with open('../monoisotopic-mass-table.txt') as file:
    for line in file:
        line = line.split()
        mass_table[line[0]] = float(line[1])

print(mass_table)
```

```python
file = open('rosalind_prtm.txt', 'r')
protein = ''
for line in file:
    line = line.strip()
    protein += line

# Adds up the masses of amino acids using mass table
mass = 0.0
for aa in protein:
    mass += mass_table[aa]

print(round(mass, 3))
```
