# Open reading frame

Problem: Find every open reading frame in a sequence of DNA and convert to protein

```
# Function that takes file and removes /t
def get_codon_dict(filename):
    file = open(filename)

    aa_dict = {}

    # Removes new line and tab. 'line' is a list of the codon and AA names. Add to dictionary aa_dict
    for line in file.readlines():
        line = line.strip()
        line = line.split('\t')
        aa_dict[line[0]] = {'1-letter': line[1], '3-letter': line[2], 'name': line[3]}

    file.close()
    return aa_dict
```

```
aa_dict = get_codon_dict('genetic_code.tsv')
print(aa_dict)
```

```
dna_to_aa_codons = {}

for item in aa_dict:
    aa_to_add = aa_dict[item]['1-letter']
    item = item.replace('U', 'T')

    dna_to_aa_codons[item] = aa_to_add

print(dna_to_aa_codons)
```

```
sequence = 'AGCCATGTAGCTAACTCAGGTTACATGGGGATGACCCCGCGACTTGGATTAGAGTCTCTTTTGGAATAAGCCTGAATGATCCGAGTAGCATCTCAG'

def find_start_codon_positions(sequence):

    start_codon_positions = []
    for i in range(len(sequence)):

        if sequence[i:i+3] == 'ATG':
            start_codon_positions.append(i)

    return start_codon_positions
```

```
start_codon_positions = find_start_codon_positions(sequence)
```

```
first_start_codon = sequence.find('ATG')

sequence_from_start = sequence[first_start_codon:]

print(sequence_from_start)

a = sequence_from_start.split('ATG')

print(a)
```

```
counter = 0

protein = ''
possible_proteins = []

for position in start_codon_positions:
    while counter < len(sequence):
        codon_to_check = sequence[position+counter:position+counter+3]
        if codon_to_check == 'TAA' or codon_to_check == 'TAG' or codon_to_check == 'TGA':
            possible_proteins.append(protein)
            protein = ''
            counter = 0
            break
        else:
            protein += dna_to_aa_codons[codon_to_check]
            counter += 3

print(possible_proteins)
```
