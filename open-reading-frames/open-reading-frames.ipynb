# Open reading frame

Problem: Find every open reading frame in a sequence of DNA and convert to protein

Get amino acid dictionary function

```
# Function that takes file and removes /t
def get_codon_dict(filename):
    file = open(filename)

    aa_dict = {}

    # Removes new line and tab. 'line' is a list of the codon and AA names. Add to dictionary aa_dict
    for line in file.readlines():
        line = line.strip()
        line = line.split('\t')
        aa_dict[line[0]] = {'1-letter': line[1], '3-letter': line[2], 'name': line[3]}

    file.close()
    dna_to_aa_codons = {}
    for item in aa_dict:
        aa_to_add = aa_dict[item]['1-letter']
        item = item.replace('U', 'T')

        dna_to_aa_codons[item] = aa_to_add

    return dna_to_aa_codons
```

```
aa_dict = get_codon_dict('genetic_code.tsv')
```

```
sequence_file = open('rosalind_orf.txt', 'r')
sequence = ''
for line in sequence_file:
    line = line.strip()
    if line[0] != '>':
        sequence += line
```

```
# Finds all the start positions in the sequence and adds them to a list
def find_start_codon_positions(sequence):

    start_codon_positions = []
    for i in range(len(sequence)):

        if sequence[i:i+3] == 'ATG':
            start_codon_positions.append(i)

    return start_codon_positions
```

```
start_codon_positions = find_start_codon_positions(sequence)
```

```

def possible_proteins(sequence, start_codon_positions):
    protein = ''
    possible_proteins = []

    for start_pos in start_codon_positions:
        # Once it goes to the start position, it then iterates through each
        # triplet starting from the start codon
        for i in range(start_pos, len(sequence), 3):
            current_codon = sequence[i:i+3]
            if current_codon == 'TAA' or current_codon == 'TAG' or current_codon == 'TGA':
                possible_proteins.append(protein)
                protein = ''
                break
            # Solves error where final codon in sequence is less than 3
            # Meaning we've reached beyond the bounds of current sequence
            elif len(current_codon) < 3:
                break
            else:
                protein += aa_dict[current_codon]
    return possible_proteins


coding_proteins = possible_proteins(sequence, start_codon_positions)
```

```
def get_non_coding_strand(sequence):
    complimentary_sequence = ''
    codon_dict = {'A': 'T',
                  'T': 'A',
                  'C': 'G',
                  'G': 'C'}

    for base in sequence:
        complimentary_sequence += codon_dict[base]

    non_coding_strand = complimentary_sequence[::-1]

    return non_coding_strand

non_coding_sequence = get_non_coding_strand(sequence)

reverse_start_codons = find_start_codon_positions(non_coding_sequence)
reverse_proteins = possible_proteins(non_coding_sequence, reverse_start_codons)
```

```
total_proteins = coding_proteins + reverse_proteins
total_proteins = list(dict.fromkeys(total_proteins))

for protein in total_proteins:
    print(protein)

```
