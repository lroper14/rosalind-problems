# Genome Assembly

```python
import time

from IPython.core.completer import position_to_cursor
import time
start = time.perf_counter()
```

```python
from useful_functions import get_seq_list

seqs = get_seq_list('rosalind_long.txt')

assembled_sequence = seqs.pop(0)
```

```python
# Function to get all the substrings such as in ATCG: A, AT, ATC, ATCG, TCG, CG, G
def get_all_substrings(sequence):

    substrings = []
    for i in range(1,len(sequence)):
        substrings.append(sequence[0:i])

    for i in range(0,len(sequence)):
        substrings.append(sequence[i:])

    substrings.sort(key=len, reverse=True)

    return substrings

substrings = get_all_substrings(assembled_sequence)
```

```python
# The idea of this function is to find where the longest common substring is within the rest of the list of sequences
# then to add the head or tail of that string to the start or end of the final assembled sequence
def find_bases_to_add(substrings):
    stop = False
    # Starts with each substring and goes through each of the sequences, checking if the largest one is in there, starting with the largest to speed things up
    for substring in substrings:
        for seq in seqs:
            if substring in seq:

                position = seq.find(substring)
                # Then these conditional statements add to the start or end of the sequence
                if position == 0:
                    sequence = assembled_sequence + seq[len(substring):]

                else:
                    sequence = seq[:position] + assembled_sequence

                stop = True

                # Important to remove the sequences that have been previously checked
                seqs.pop(seqs.index(seq))
                break

        if stop:
            break



    return sequence


```

```python
# Using a counter because the sequence list gets popped during the process
counter = 0
while counter < len(seqs):
    substrings = get_all_substrings(assembled_sequence)
    assembled_sequence = find_bases_to_add(substrings)
```
