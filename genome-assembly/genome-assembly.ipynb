# Genome Assembly

```python
import time
```

```python
from useful_functions import get_seq_list

seqs = get_seq_list('rosalind_long.txt')
print(seqs)
```

```python

assembled_sequence = seqs[0]
substrings = []

time_get_substrings = 0
time_find_longest_common = 0

for iter, seq in enumerate(seqs):
    start = time.perf_counter()
    for i in range(len(seq)):
#============================================================================
        timestart_get_substrings = time.perf_counter()
#============================================================================
        # Then the next loop goes through all the possible substrings starting from that base
        for j in range(i +1, len(seq) +1):
            # substring stores each substring formed starting from base i going up to, but not including, base j
            substring = seq[i:j]
            substrings.append(substring)

#============================================================================
        timeend_get_substrings = time.perf_counter()
        time_get_substrings = timeend_get_substrings-timestart_get_substrings
#============================================================================

    substrings.sort(key=len, reverse = True)

    longest_common = ''
#============================================================================
    timestart_find_longest_common = time.perf_counter()
#============================================================================


    for substring in substrings:
        if substring in assembled_sequence:
            if len(substring) > len(longest_common):
                longest_common = substring
            else:
                break

#============================================================================
    timeend_find_longest_common = time.perf_counter()
    time_find_longest_common = timeend_find_longest_common - timestart_find_longest_common
#============================================================================


    substring_pos = seq.find(longest_common)


    before = seq[:substring_pos]
    after = seq[substring_pos+len(longest_common):]

    if before == '':
        assembled_sequence += after
    else:
        assembled_sequence = before + assembled_sequence

    substrings = []
    longest_common = ''

#============================================================================
    end = time.perf_counter()
#============================================================================

    print(f'Total time for iteration {iter}: {end - start:.4f}s')
    print(f'Time to get substrings: {time_get_substrings}s')
    print(f'Time find longest common substrings: {time_find_longest_common}s')
    print()




print(assembled_sequence)


```

```python
with open('sequence.txt', 'w') as f:
    f.write(assembled_sequence)

with open("sequence.txt", "r", encoding="utf-8") as f:
    print(f.read())
```
