# Genome Assembly

```python
from useful_functions import get_seq_list

seqs = get_seq_list('assembly.txt')
print(seqs)
```

```python

assembled_sequence = seqs[0]
substrings = []

for seq in seqs:
    for i in range(len(seq)):
        # Then the next loop goes through all the possible substrings starting from that base
        for j in range(i +1, len(seq) +1):
            # substring stores each substring formed starting from base i going up to, but not including, base j
            substring = seq[i:j]
            substrings.append(substring)

    longest_common = ''
    for substring in substrings:
        if substring in assembled_sequence:
            if len(substring) > len(longest_common):
                longest_common = substring

    print(longest_common)
    substring_pos = seq.find(longest_common)

    print(substring_pos)

    before = seq[:substring_pos]
    after = seq[substring_pos+len(longest_common):]
    print(after)

    if before == '':
        assembled_sequence += after
    else:
        assembled_sequence = before + assembled_sequence

    substrings = []
    longest_common = ''

print(assembled_sequence)
```
